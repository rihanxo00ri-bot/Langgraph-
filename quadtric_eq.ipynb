# quadtric equation conditional workflow


```python
from langgraph.graph import StateGraph, START,END
from typing import TypedDict, Literal
```

```python
#define state
class QuadState(TypedDict):
    a:int
    b:int
    c:int
    d:float
    result:str
    equation:str
    
    
```

```python
#crete node

def show_equation(state:QuadState)->QuadState:
    """It take the variable and convert to Equation"""
    equation = f'{state['a']}x2 + {state['b']}x +{state['c']}'

    return {'equation':equation}
```

```python
# calculate another node with calcuate discrimant

def dis_eq(state:QuadState)->QuadState:
    """It calcualte the discriminant"""
    d=state['b']**2 -(4*state['a']*state['c'])

    return {'d':d}
```

```python
def real_roots(state: QuadState):

    root1 = (-state["b"] + state["d"]**0.5)/(2*state["a"])
    root2 = (-state["b"] - state["d"]**0.5)/(2*state["a"])

    result = f'The roots are {root1} and {root2}'

    return {'result': result}

def repeated_roots(state: QuadState):

    root = (-state["b"])/(2*state["a"])

    result = f'Only repeating root is {root}'

    return {'result': result}

def no_real_roots(state: QuadState):

    result = f'No real roots'

    return {'result': result}
```

```python
#define graph

graph = StateGraph(QuadState)

# add node
graph.add_node('show_equation', show_equation)
graph.add_node('dis_eq',dis_eq)

#add edges
graph.add_edge(START,'show_equation')
graph.add_edge('show_equation','dis_eq')
graph.add_edge('dis_eq',END)


graph.compile()

```
