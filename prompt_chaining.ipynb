PROMPT CHAINING WORKFLOW


```python
from langgraph.graph import StateGraph,  START,END
from typing import TypedDict
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
#load env
load_dotenv()
#load model
model = ChatGoogleGenerativeAI(model='gemini-3.6-flash')
```

```python
#define state

class Prompt(TypedDict):
    title:str
    outline:str
    content:str
    
```

```python
# outline gernator

def outliner(state:Prompt)-> Prompt:
    """This function create a title to outline"""
    # take the title
    title=state['title']

    #create the outine
  
    prompt = f'write a outline which is in professional format on this title ->{title}'
    outline = model.invoke(prompt).content

    # update the state
    state['outline']= outline

    return state
```

```python
#blog writer

def Blog_writer(state:Prompt)-> Prompt:
    """THis function write the blogs"""
    title = state['title']
    outline=state['outline']

    prompt = f"Write a proper deatil blog on this title ->{title} using this outline ->{outline}"

    #invoke the model
    content = model.invoke(prompt).content
    state['content']=content

    return state
```

```python
#define graph
graph= StateGraph(Prompt)
#add node
graph.add_node('create_out', outliner)
graph.add_node('Blog_writer',Blog_writer)

#add edges
graph.add_edge(START,'create_out')
graph.add_edge('create_out','Blog_writer')
graph.add_edge('Blog_writer',END)

#complie graph





```

```python
workflow = graph.compile()

init_state = {'title': 'backpropogation'}
final_answer =workflow.invoke(init_state)
print(final_answer)
```
